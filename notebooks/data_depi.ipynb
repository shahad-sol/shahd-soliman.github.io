```python
#loop
def load_dataset(path: str) -> pd.DataFrame:
    encodings = ['utf-8', 'latin-1', 'ISO-8859-1']
    for encoding in encodings:
        try:
            df = pd.read_csv(path, encoding=encoding)
            print(f"Dataset loaded successfully using {encoding}")
            return 
        except UnicodeDecodeError:
            print(f"Failed to load using {encoding}")
    raise ValueError("Could not load the dataset with the available encodings.")
```

```python
#cleaning starts from here 'محاضرة 3'

def handle_missing_values(data: pd.DataFrame , numeric_strategy="median", categorical_strategy="mode")->pd.DataFrame:
    numeric_cols=data.select_dtypes(include=np.number).columns
    categolical_cols=data.select_dtypes(include='str').columns
    # in vs code object doesnt work only 'str'

    for col in numeric_cols:
        if data[col].isnull().sum()>0:
            fill_value=data[col].median() if numeric_strategy == 'median' else data[col].mean()
            data[col] = data[col].fillna(fill_value)
    for col in categolical_cols:
        if data[col].isnull().sum()>0:
             fill_value = data[col].mode()[0]
             data[col]= data[col].fillna(fill_value)
    print('we have removed the null thing')             
    return data            


     
```

```python
def generate_kpi_summary(data):
    kpis = {
        'Total Sales': data['Sales'].sum(),
        'Total Profit': data['Profit'].sum(),
        'Overall Profit Margin %': (data['Profit'].sum() / data['Sales'].sum()) * 100,
        'Average Order Value': data['Sales'].sum() / data['Order ID'].nunique(),
        'Average Discount': data['Discount'].mean(),
    }
    for k, v in kpis.items():
        print(f"{k}: {round(v, 2)}")
    return kpis
```

```python
pd.DataFrame(list(generate_kpi_summary(df).items()), columns=['KPI', 'Value']).to_csv('kpi_summary.csv', index=False)
```
